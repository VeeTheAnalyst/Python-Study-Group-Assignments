# Task 1: Secure Login System

```python
Username_correct = True
Password_correct = True
Account_locked = False

if Username_correct:
    print("Access Granted")
else:
    print("Access Denied")

# using the 'and' operator
if Username_correct and Password_correct:
    print("Access Granted")
else:
    print("Access Denied")

#using the 'not' operator
not Account_locked
```

```python
if not Account_locked:
    print("Access Granted")
else:
    print("Access Denied")

# combining multiple condition
if Username_correct and Password_correct and not Account_locked:
    print("Access Granted")
else:
    print("Acess Denied")
```

```python
# challenge
Wrong_password = True
Locked_account = False
Incorrect_username = True

if Wrong_password:
    print("Access Denied")
else:
    print("Access Granted")
    
# --------------
if Incorrect_username:
    print("Access Denied")
else:
    print("Access Granted")
    
# -----------------
if Locked_account:
    print("Access Denied")
else:
    print("Access Granted")
    
#-----------------
if Wrong_password and Incorrect_username and not Locked_account:
    print("Access Denied")
else:
    print("Acess Granted")
```

# Task 3: Smart Server Health Monitor

```python
#CPU threshold is 90%
#Memory usage threshold is 90%
#Disk usage threshold is 95%

CPU_usage = 75
Memory_usage = 92
Disk_usage = 60

if CPU_usage > 90 or Memory_usage > 90 or Disk_usage > 95:
    print("Critical Alert!")
else:
    print("System Healthy")
```
