# Exercise1

```python
First_name = "Grace" 
Last_name = "Johnson"
Age = 34 
Blood_group = "O+"
Department = "Cardiology"

# Using String Concatenation(+)

print("==========CAREPLUS HOSPITAL==========")
print("Patient Name:" + First_name + " " + Last_name)
print("Age:" + str(Age) + " " + "Years")
print("Blood Group:" + Blood_group)
print("Department:" + Department)
print("======================================")
```

# Bonus Challenge

```python
# Using .format()

print("==========CAREPLUS HOSPITAL==========")
print("Patient Name: {} {}" .format(First_name, Last_name))
print("Age: {}" .format(Age))
print("Blood Group: {}" .format(Blood_group))
print("Department: {}" .format(Department))
print("======================================")
```

```python
# Using f-string
Output = f'''
==========CAREPLUS HOSPITAL==========
Patient Name: {First_name} {Last_name}
Age: {Age}
Blood Group: {Blood_group}
Department: {Department}
======================================'''

print(Output)
```

```python
# Using Uppercase for the Patient's full name
Full_name = (First_name + " " + Last_name) .upper()
Output = f'''
==========CAREPLUS HOSPITAL==========
Patient Name: {Full_name}
Age: {Age}
Blood Group: {Blood_group}
Department: {Department}
======================================'''

print(Output)
```

# Exercise2

```python
# Creating variable
Log = "  LOGIN SUCCESSFUL FROM ATM-045 ON 2026-07-27  "

# To print the origiinal log

print(Log)

# To remove all leading and trailing spaces
Cleaned_log = Log.strip()
print(Cleaned_log)

# To convert to lower_case
Lowercase_log = Cleaned_log .lower()
print(Lowercase_log)
```

```python
# To find the position of the word "SUCCESSFUL"
Position = Cleaned_log .find("SUCCESSFUL")
print(Position)

# To replace "SUCCESSFUL" with "VERIFIED" and display the updated log message
Updated_log = Cleaned_log .replace("SUCCESSFUL", "VERIFIED")
print(Updated_log)
```

# Bonus Challenge

```python
# Number of characters before cleaning
print(len(Log))

# Number of characters after cleaning
print(len(Cleaned_log))

# Number of words in the cleaned log
print(len(Cleaned_log.split()))
```

# Exercise3

```python
# Creating a variable
Movie_title = "The Last Kingdom Returns"

# To print the first character
print( Movie_title[0])

# To print the last character
print(Movie_title[-1])

# To print the first word only
print(Movie_title[:3])

# To print the last word only
print(Movie_title[17:])

# To find the starting position of the word "Returns"
print(Movie_title.find("Returns"))

# To split the movie title into individual words
print(Movie_title.split())
```
