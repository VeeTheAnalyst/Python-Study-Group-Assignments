# Task 2: Payroll System

```python
Employee_name = "David Musa"
Department = "Finance"
Basic_salary = 185000
Bonus = 35000
Tax_rate = 7.5

Gross_salary = Basic_salary + Bonus
Tax_amount = Gross_salary * (Tax_rate / 100)
Net_salary = Gross_salary - Tax_amount

Output = f'''
===================================
      PAYROLL SUMMARY
===================================

Employee : {Employee_name}
Department : {Department}
Basic Salary : #{Basic_salary:.2f}
Bonus : #{Bonus:.2f}
Gross Salary : #{Gross_salary:.2f}
Tax Amount(7.5%) : #{Tax_amount:.2f}
Net Salary : #{Net_salary:.2f}

===================================='''

print(f"#{Gross_salary:.2f}")
print(f"#{Tax_amount:.2f}")
print(f"#{Net_salary:.2f}")
print(Output)
```

# Task 3: E-commerce Order Receipt

```python
Customer_name = "Amina Bello"
Product_name = "Wireless Headphones"
Quantity = 3
Unit_price = 27500
Total_cost = Quantity * Unit_price 

Receipt = f'''
=======================================
      ONLINE STORE RECEIPT
=======================================

Customer : {Customer_name}
Product : {Product_name}
Quantity : {Quantity}
Unit Price : #{Unit_price:.2f}
Total Cost : #{Total_cost:.2f}

Thank you for shopping with us!

=========================================
'''

print(Total_cost)
print(Receipt)
```

```python
#10% discount if customer buys more than 2 items
#since item is 3
Discount = Total_cost * (10 / 100)
Final_amount = Total_cost - Discount

print(f"#{Discount:.2f}")
print(f"#{Final_amount:.2f}")
```
