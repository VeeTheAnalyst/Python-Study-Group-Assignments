# Exercise1: Employee Record Information Extractor(HR)

```python
# Creating a variable
Employee_record = "EMP-HR-2026-004512"
print(Employee_record)

# Using string slicing
print("Employee Prefix:", Employee_record[:3])
print("Department:", Employee_record[4:6])
print("Employment Year:", Employee_record[7:11])
print("Employee Number:", Employee_record[-6:])
print("First 10 Characters:", Employee_record[:10])
print("Last 6 Characters:", Employee_record[-6:])
print("Every Second Character:", Employee_record[ ::2])
print("Every Third Character:", Employee_record[ ::3])
print("Reverse:", Employee_record[::-1])
```

# Exercise 2: Banking Card Security Formatter

```python
# Creating variable
Card_number = "5489123476549876"

print("First 4 Digits:", Card_number[:4])
print("Last 4 Digits:", Card_number[-4:])
print("Middle 8 Digits:", Card_number[4:12])
print("Every Second Digit:", Card_number[::2])
print("Every Third Digit:", Card_number[::3])
print("Card Number In Reverse:", Card_number[::-1])
print("Last 8 Digits In Reverse:", Card_number[:7:-1])
```

# Exercise 3: Customer Information Processor (Retail Analytics)

```python
# Creating variable
Customer_record = "CUS-LAG-2026-000451"
print(Customer_record)

# Using string slicing
print("Customer Prefix:", Customer_record[:3])
print("City Code:", Customer_record[4:7])
print("Registration Year:", Customer_record[8:12])
print("Customer Number:", Customer_record[-6:])
print("First 8 Characters:", Customer_record[:8])
print("Last 5 Characters:", Customer_record[-5:])
print("Every Second Character:", Customer_record[::2])
print("Every Third Character:", Customer_record[::3])
print("Reverse:", Customer_record[::-1])

# Creating another varible
Customer_name = "Christopher Johnson"
print("First Name:", Customer_name[:11])
print("Last Name:", Customer_name[12:])
print("First 5 Letters:", Customer_name[:5])
print("Last 7 Letters:", Customer_name[-7:])
print("Reverse Name:", Customer_name[::-1])
```
